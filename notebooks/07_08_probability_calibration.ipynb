---

## The Core Question of Phase 2

> **"When PredictGuard predicts a 70% probability of failure for a machine over the next 24 hours, do approximately 70% of those machine-hours actually fail?"**

High ROC-AUC (0.999) or PR-AUC (0.973) does **NOT** guarantee that predicted probabilities are calibrated or trustworthy. Class-imbalanced tree models trained with `scale_pos_weight` often push probabilities toward extreme values, producing **over-confident** or **under-confident** risk estimates.

Phase 2 evaluates raw probability quality (**Stage 7**) and improves probability calibration (**Stage 8**) using post-hoc Isotonic Regression & Sigmoid (Platt Scaling) fitted strictly on the Development set.

---
## 0. Setup & Imports

```python
import logging
import sys
import time
from pathlib import Path
import pandas as pd
import numpy as np
import joblib
import matplotlib.pyplot as plt

project_root = Path.cwd().parent
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from src.calibration import (
    CalibrationEvaluator, ProbabilityCalibrator, ReliabilityAnalyzer,
    plot_calibration_dashboard, plot_reliability_diagrams, plot_probability_histograms,
    generate_calibration_report,
)

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s | %(levelname)-8s | %(message)s",
    datefmt="%H:%M:%S",
    handlers=[logging.StreamHandler(sys.stdout)],
)

PROCESSED_DIR = project_root / "data" / "processed"
REPORTS_DIR   = project_root / "reports"
FIGURES_DIR   = REPORTS_DIR / "figures"
MODELS_DIR    = project_root / "models"
```

---
## 1. Load Parquets & Trained Models

```python
dev_df  = pd.read_parquet(PROCESSED_DIR / "development.parquet")
test_df = pd.read_parquet(PROCESSED_DIR / "test.parquet")

non_feature_cols = {"datetime", "machineID", "y_failure", "failure_component", "time_to_failure_hours"}
feature_cols = [c for c in dev_df.columns if c not in non_feature_cols]

X_dev, y_dev, groups_dev = dev_df[feature_cols].select_dtypes(include=[np.number]).fillna(0.0), dev_df["y_failure"].astype(int), dev_df["machineID"]
X_test, y_test, groups_test = test_df[feature_cols].select_dtypes(include=[np.number]).fillna(0.0), test_df["y_failure"].astype(int), test_df["machineID"]

models = {
    "XGBoost": joblib.load(MODELS_DIR / "xgboost.pkl"),
    "Random Forest": joblib.load(MODELS_DIR / "random_forest.pkl"),
    "Logistic Regression": joblib.load(MODELS_DIR / "logistic_regression.pkl"),
}
print(f"Loaded {len(models)} models and datasets (Dev: {len(dev_df):,}, Test: {len(test_df):,})")
```

---
## 2. Stage 7: Calibration Evaluation of Raw Probabilities

```python
evaluator = CalibrationEvaluator(n_bins=10)
raw_results = {}
raw_records = []

for m_name, model in models.items():
    y_prob = model.predict_proba(X_test)[:, 1]
    res = evaluator.evaluate(m_name, y_test.values, y_prob)
    res["y_prob"] = y_prob
    raw_results[m_name] = res
    raw_records.append({
        "Model": m_name, "Brier Score": res["brier_score"], "ECE": res["ece"],
        "MCE": res["mce"], "Log Loss": res["log_loss"], "Status": res["status"]
    })

raw_df = pd.DataFrame(raw_records).sort_values("ECE").reset_index(drop=True)
display(raw_df)
```

---
## 3. Stage 8: Post-Hoc Calibration (Isotonic vs Sigmoid)

> Calibration is fitted **strictly on the 80 Development machines** using `StratifiedGroupKFold(5)` — final test machines remain untouched.

```python
calibrator = ProbabilityCalibrator(n_folds=5, random_seed=42)
xgb_raw = models["XGBoost"]

# Isotonic
iso_model = calibrator.fit_calibrated_model(xgb_raw, X_dev, y_dev, groups_dev, method="isotonic")
iso_prob  = iso_model.predict_proba(X_test.values)[:, 1]
iso_eval  = evaluator.evaluate("XGBoost (Isotonic)", y_test.values, iso_prob)
iso_eval["y_prob"] = iso_prob
iso_eval["method"] = "Isotonic"

# Sigmoid
sig_model = calibrator.fit_calibrated_model(xgb_raw, X_dev, y_dev, groups_dev, method="sigmoid")
sig_prob  = sig_model.predict_proba(X_test.values)[:, 1]
sig_eval  = evaluator.evaluate("XGBoost (Sigmoid)", y_test.values, sig_prob)
sig_eval["y_prob"] = sig_prob
sig_eval["method"] = "Sigmoid"

raw_xgb_eval = raw_results["XGBoost"]
raw_xgb_eval["method"] = "None (Raw)"

comp_df = pd.DataFrame([
    {"Variant": "Raw XGBoost",        "Method": "None",     "Brier": raw_xgb_eval["brier_score"], "ECE": raw_xgb_eval["ece"], "MCE": raw_xgb_eval["mce"], "PR-AUC": raw_xgb_eval["pr_auc"]},
    {"Variant": "Calibrated XGBoost", "Method": "Isotonic", "Brier": iso_eval["brier_score"],     "ECE": iso_eval["ece"],     "MCE": iso_eval["mce"],     "PR-AUC": iso_eval["pr_auc"]},
    {"Variant": "Calibrated XGBoost", "Method": "Sigmoid",  "Brier": sig_eval["brier_score"],     "ECE": sig_eval["ece"],     "MCE": sig_eval["mce"],     "PR-AUC": sig_eval["pr_auc"]},
])
display(comp_df)
```

---
## 4. Probability Audit Table (Raw vs Calibrated Risk)

```python
best_cal_eval = iso_eval if iso_eval["ece"] <= sig_eval["ece"] else sig_eval
audit_df = ReliabilityAnalyzer.audit_probabilities(
    y_true=y_test, raw_probs=raw_xgb_eval["y_prob"], calibrated_probs=best_cal_eval["y_prob"],
    machine_ids=test_df["machineID"], timestamps=test_df["datetime"], n_samples=15,
)
display(audit_df)
```

---
## 5. Visualizations & Figure Generation

```python
plot_calibration_dashboard(
    eval_raw_models=raw_results, raw_eval=raw_xgb_eval, iso_eval=iso_eval, sig_eval=sig_eval,
    y_test=y_test.values, raw_probs=raw_xgb_eval["y_prob"], calibrated_probs=best_cal_eval["y_prob"],
    figures_dir=FIGURES_DIR,
)
plot_reliability_diagrams({"Raw XGBoost": raw_xgb_eval, "Isotonic XGBoost": iso_eval, "Sigmoid XGBoost": sig_eval}, FIGURES_DIR)
plot_probability_histograms(raw_xgb_eval["y_prob"], best_cal_eval["y_prob"], FIGURES_DIR)
print("✅ Figures saved under reports/figures/")
```
